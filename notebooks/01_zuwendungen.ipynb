```python
maske_vr = jahr.loc[:, "Zuwendungsempfänger"] == "Vereinsregister"
jahr.loc[maske_vr, ["Zuwendungszweck", "Behörde", "netto"]]
```

```python
jahr.loc[maske_vr, "Zuwendungsempfänger"] = "Vereinsregister " + jahr.loc[maske_vr, "INEZ-Nummer"]
```

```python
#C2 — pro Empfänger summieren (Schreibweisen zusammengeführt):

jahr.loc[:, "empf_schluessel"] = (jahr.loc[:, "Zuwendungsempfänger"]
                                  .str.lower()
                                  .str.replace(r"[^a-z0-9]", "", regex=True))

je_schreibweise = (jahr.groupby(["empf_schluessel", "Zuwendungsempfänger"])["netto"]
                   .sum()
                   .reset_index()
                   .sort_values("netto", ascending=False))
anzeigename = (je_schreibweise.drop_duplicates("empf_schluessel")
               .set_index("empf_schluessel")
               .loc[:, "Zuwendungsempfänger"])

empfaenger = jahr.groupby("empf_schluessel")["netto"].sum().sort_values(ascending=False)
empfaenger.index = empfaenger.index.map(anzeigename).rename("Zuwendungsempfänger")
empfaenger.head(10)
```

## Befund: nicht-städtische Empfänger

Die Top-50-Empfänger wurden gegen den Beteiligungsbericht der FHH geprüft (städtisch / gemischt-öffentlich / nicht-städtisch; unklare Fälle vorsichtshalber als nicht-städtisch gewertet).

| Gruppe | Anteil an 2,62 Mrd € |
|---|---|
| Städtisch | 80,3 % |
| Gemischt-öffentlich | 1,3 % |
| Nicht-städtisch | 18,4 % (481 Mio €) |

Unter den 1.945 nicht-städtischen Empfängern ist das Geld deutlich breiter gestreut als im Gesamtbild: Die Top 20 vereinen 37,4 % (statt 82,5 % gesamt), und 45 Empfänger brauchen für die Hälfte der Summe – nicht eine einzelne AG.

```python
schluessel = (empfaenger.index.to_series()
              .str.lower()
              .str.replace(r"[^a-z0-9]", "", regex=True))
maske_dopplung = schluessel.duplicated(keep=False)
print(len(empfaenger), "Empfänger nach dem Zusammenführen")
print(maske_dopplung.sum(), "Namen noch doppelt")
empfaenger.index[maske_dopplung]
```

```python
letztes_datum = df.groupby("INEZ-Nummer")["Bescheiddatum"].max()
df.loc[:, "letztes_datum"] = df.loc[:, "INEZ-Nummer"].map(letztes_datum)
maske_letzt = df.loc[:, "Bescheiddatum"] == df.loc[:, "letztes_datum"]
je_vorgang = df.loc[maske_letzt, :].groupby("INEZ-Nummer").size()
(je_vorgang > 1).sum(), len(je_vorgang)

# Bescheide amselben Tag. Zählt, bei wie vielen Vorgängen das späteste Bescheiddatum mehrfach vorkommt. Nur dort entscheidet die zufällige Zeilenreihenfolge, welcher Bescheid gewinnt.
```

```python
for spalte in ["INEZ-Nummer", "Zuwendungsempfänger",
               "Summe von Zuwendungssumme", "Bescheiddatum", "Summe von Rückforderungen"]:
    print(spalte, "->", df.loc[:, spalte].isna().sum(), "fehlend von", len(df))
```

```python
strittig = je_vorgang[je_vorgang > 1].index
maske_strittig = df.loc[:, "INEZ-Nummer"].isin(strittig) & maske_letzt

spannen = (df.loc[maske_strittig, :]
           .groupby("INEZ-Nummer")[["Summe von Zuwendungssumme",
                                    "Summe von Rückforderungen"]]
           .nunique())
(spannen > 1).any(axis=1).sum()

#Für die 154 strittigen Vorgänge wird gezählt, wie viele verschiedene Werte in den beiden Geldspalten vorkommen. 
# Ist es überall genau einer, ist die Auswahl folgenlos. 
#any(axis=1) heißt: reicht schon, wenn eine der beiden Spalten uneins ist.
```

```python
strittige_ids = spannen.index[(spannen > 1).any(axis=1)]
maske_j = jahr.loc[:, "INEZ-Nummer"].isin(strittige_ids)
maske_j.sum(), jahr.loc[maske_j, "netto"].sum(), jahr.loc[:, "netto"].sum()

#zählt, wie viele der 32 strittigen Vorgänge in deinem 2025-Datensatz stecken, 
# und stellt deren Geld der Gesamtsumme von 2,62 Mrd € gegenüber.

#Von den 32 uneindeutigen Vorgängen liegt genau einer in deinem 2025-Datensatz, mit 13.425 € von 2,616 Mrd €. Das sind 0,0005 %.

#Kein zweites Sortierkriterium nötig. Ein Satz ins Notebook, fertig: geprüft, ein Fall, Größenordnung fünf Zehntausendstel Prozent, ohne Einfluss auf das Ergebnis.
```

```python
for name, daten in [("Rohzeilen", df), ("Vorgänge", vorgaenge), ("nur 2025", jahr)]:
    fehlend = daten.loc[:, "Summe von Rückforderungen"].isna().sum()
    print(f"{name}: {fehlend} von {len(daten)} = {fehlend/len(daten)*100:.1f} %")
```
