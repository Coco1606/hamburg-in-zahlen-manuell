```python
vorgaenge = df.sort_values(["INEZ-Nummer", "Bescheiddatum"]).drop_duplicates(subset="INEZ-Nummer", keep="last").copy()
```

```python
#C1 — auf 2025 einschränken:

maske_2025 = vorgaenge.loc[:, "erstjahr"] == 2025
jahr = vorgaenge.loc[maske_2025, :].copy()
len(jahr), jahr.loc[:, "netto"].sum()
```
